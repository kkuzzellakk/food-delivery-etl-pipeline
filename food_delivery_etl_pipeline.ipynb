```python
import pandas as pd
df = pd.read_csv('/talabat_enhanced_orders.csv')
print(f"Початкова кількість рядків: {len(df)}")
duplicates = df.duplicated(subset=['Order_ID']).sum()
nulls = df['Order_ID'].isnull().sum()
print(f"Знайдено дублікатів замовлень: {duplicates}")
print(f"Порожніх Order_ID: {nulls}")
df_clean = df.drop_duplicates(subset=['Order_ID'])
orders_fact = df_clean[['Order_ID', 'User_ID', 'Restaurant_ID', 'Item_Name', 'Quantity', 'Total_Price', 'Order_Time', 'Order_Status']]
restaurants_dim = df_clean[['Restaurant_ID', 'City', 'Restaurant_Lat', 'Restaurant_Lon']].drop_duplicates(subset=['Restaurant_ID'])
orders_fact.to_csv('orders_fact.csv', index=False)
restaurants_dim.to_csv('restaurants_dim.csv', index=False)
print(f"Готово! У таблиці фактів {len(orders_fact)} рядків, у довіднику ресторанів — {len(restaurants_dim)} закладів.")
```
